```python
import os
import json
import pandas as pd
from pathlib import Path

# set the root directory containing all model subdirectories
root_dir = Path("~/p-dsgt_clef2025-0/shared/birdclef/models/2025/v2").expanduser()

# define the list to store rows
rows = []

# loop through each model directory
for model_name in os.listdir(root_dir):
    model_dir = os.path.join(root_dir, model_name)
    results_path = os.path.join(model_dir, "evaluation_results.json")

    if not os.path.isfile(results_path):
        continue

    with open(results_path, "r") as f:
        results = json.load(f)

    for split in ["training_metrics", "validation_metrics"]:
        row = {
            "model_name": results["model_name"],
            "split": "train" if split == "training_metrics" else "val",
            "accuracy": results[split].get("accuracy"),
            "f1_macro": results[split].get("f1_macro"),
            "f1_micro": results[split].get("f1_micro"),
            "f1_weighted": results[split].get("f1_weighted"),
            "recall": results[split].get("recall_macro"),
            "precision_macro": results[split].get("precision_macro"),
            "auc_roc_macro": results[split].get("auc_macro"),
            "auc_roc_weighted": results[split].get("auc_weighted"),
        }
        rows.append(row)

# create a DataFrame
df = pd.DataFrame(rows)

# Save to CSV
output_csv_path = os.path.join(root_dir, "compiled_evaluation_results.csv")
df.to_csv(output_csv_path, index=False)
print(f"Compiled evaluation results saved to: {output_csv_path}")
df.head()
```
